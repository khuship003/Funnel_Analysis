```python
import pandas as pd

df = pd.read_csv("../data/cleaned/events_clean.csv")
df["event_time"] = pd.to_datetime(df["event_time"], dayfirst=True)

# First-seen week per user = their cohort
df["cohort_week"] = df.groupby("user_id")["event_time"].transform("min").dt.to_period("W")
df["activity_week"] = df["event_time"].dt.to_period("W")
df["week_number"] = (df["activity_week"] - df["cohort_week"]).apply(lambda x: x.n)

cohort_counts = df.groupby(["cohort_week","week_number"])["user_id"].nunique().unstack()
cohort_size = cohort_counts.iloc[:, 0]
retention = cohort_counts.divide(cohort_size, axis=0).round(3)

retention.to_csv("../data/cleaned/retention_matrix.csv")
retention
```
